```python
# Importing Libraries

import pandas as pd
import requests
import matplotlib.pyplot as plt
```

```python
url="https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
response=requests.get(url)
```

```python
data=response.json()
```

```python
df=pd.DataFrame(data["items"])
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
# Selecting Required Columns

df = df[["name", "owner", "language", "stargazers_count", "forks_count", "watchers_count", "open_issues_count", "created_at", "updated_at", "license"]]
```

```python
df
```

```python
df.head()
```

```python
df.info()
```

```python
# Extracting Owner's Username

df['owner'] = pd.json_normalize(df['owner'])['login']
```

```python
# Extracting License Name

df['license'] = pd.json_normalize(df['license'])['name']
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# Handling Missng Values

df['language'] = df['language'].fillna(df['language'].mode()[0])

df['license'] = df['license'].fillna(df['license'].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Checking Duplicates

df.duplicated().sum()
```

```python
df.dtypes
```

```python
# Converting Date Columns To Datetime

df["created_at"] = pd.to_datetime(df["created_at"])

df["updated_at"] = pd.to_datetime(df["updated_at"])
```

```python
df.dtypes
```

```python
# Renaming Columns

df.rename(columns={ "stargazers_count": "stars", "forks_count": "forks", "watchers_count": "watchers", "open_issues_count": "open_issues", "created_at": "created_date", "updated_at": "updated_date"}, inplace=True)
```

```python
df.head()
```

```python
# Saving Dataframe as CSV File

df.to_csv("github_projects.csv", index=False)
```

```python
# Reading The CSV File

github_projects = pd.read_csv("github_projects.csv")
```

```python
# Seeting SQL Extension & Connection

%load_ext sql

%config SqlMagic.style = 'PLAIN_COLUMNS'

%sql sqlite:///github_projects.db
```

```python
# Saving Dataframe Directly To SQLite Database

import sqlite3

conn = sqlite3.connect("github_projects.db")

github_projects.to_sql("repositories", conn, if_exists="replace", index=False)
```

```python
# Filter repositories with more than 10,000 stars

%%sql
SELECT * FROM repositories
WHERE stars > 10000;
```

```python
# Find repositories whose name contains the word 'Machine'

%%sql
SELECT name FROM repositories
WHERE name LIKE "%Machine%" ;
```

```python
# Display the Top 10 repositories based on the number of stars

%%sql
SELECT * FROM repositories
ORDER BY stars
DESC
LIMIT 10;
```

```python
# Count the total number of repositories

%%sql
SELECT COUNT(*) AS total_repositories FROM repositories;
```

```python
# Calculate the average number of stars

%%sql
SELECT AVG(stars) AS average_stars FROM repositories;
```

```python
# Group repositories by programming language

%%sql
SELECT language, COUNT(*) AS repo_count FROM repositories
GROUP BY language;
```

```python
# Display programming languages that contain more than 5 repositories

%%sql
SELECT language, COUNT(*) AS repo_count FROM repositories
GROUP BY language
HAVING repo_count > 5;
```

```python
# Query with AND Operator

%%sql
SELECT * FROM repositories
WHERE language = "Python" AND stars > 10000;
```

```python
# Query with OR Operator

%%sql
SELECT name, stars, language FROM repositories
WHERE language = "Python" OR language = "C++";
```

```python
# Query with NOT Operator

%%sql
SELECT * FROM repositories
WHERE NOT language = "Python" ;
```

```python
top_10 = df.sort_values(by="stars", ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top_10["name"], top_10["stars"])

plt.title("Top 10 Repositories Based On Star Count")
plt.xlabel("Repository Name")
plt.ylabel("Number Of Stars")

plt.xticks(rotation=45, ha="right")


plt.show()
```

```python
yearly_counts = (df['created_date'].dt.year.value_counts().sort_index())

plt.figure(figsize=(10, 5))
plt.plot(yearly_counts.index, yearly_counts.values, marker='o')

plt.title('Repository Creation Trend by Year')
plt.xlabel('Year')
plt.ylabel('Number of Repositories')

plt.grid(True)
plt.show()
```

# **Key Findings & Insights**

Insights from Charts
 * **Peak Creation:** Most top Machine Learning projects were created between 2015 and 2018.
 * **Drop in New Projects:** Project creation dropped after 2018 because main frameworks were already built.
 * **Top Starred Repositories:** TensorFlow has the highest number of stars among all repositories.
 * **Large Star Gap:** There is a huge gap in star counts between the top 2 frameworks and the rest of the list.
 * **Steady Growth Era:** Before 2015, the creation of top ML projects was growing slowly.

General Data Insights
 * **Python Dominance:** Python is used in almost all top ML projects.
 * **Frequent Updates:** Most repositories were updated recently, showing they are active.
 * **High Engagement:** Projects with high stars also have a high number of forks.

 * **Manageable Issues:** Most popular repositories keep their open issues relatively low.
